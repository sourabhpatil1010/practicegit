## Tutorial 3 - Python Control Flows

```python
# if statement
val = float(34)
print(val)

if(val > 30):
    print("the value is greater than 30")

# if-else statement -> num is odd or even
val = 31
if(val % 2 == 0):
    print("the num is even")
else:
    print("the num is odd")
```

```python
## age form
# Nested if-else statement
age = 25
if(age < 18):
    print("you are a minor Age")
elif(age >= 18 and age <= 45):
    print ("you are mid age")
elif(age > 45 and age <= 60):
    print("you are senior age")
else:
    print("you are senior citizen")
```

```python
# nested if-else statement
age = 13
if(age < 18):
    print("you are a minor Age")
    if(age < 15):
        print("you are in a school")
    else:
        print("you are in a college")
elif(age >= 18 and age <= 45):
    print ("you are mid age")
elif(age > 45 and age <= 60):
    print("you are senior age")
else:
    print("you are senior citizen")
```

## Loops Statement -

```python
# for loop

n = 5
for i in range(n):
    if(i % 2 == 0):
        print(i)

for i in range(1, n+1):
    if(i % 2 != 0):
        print(i)
```

```python
lst = [1,2,3,4,5,6,7]
for i in lst:
    print(i ** 2)
```

```python
# find the sum of all elements in a list
lst = [1,2,3,4,5,6,7]
sum1 = 0
for i in lst:
    sum1 += i

print(sum1)
```

```python
# find the sum of even and odd numbers in a list
lst = [1,2,3,4,5,6,7]
sum_even = 0
sum_odd = 0
for i in lst:
    if(i % 2 == 0):
        sum_even += i
    else:
        sum_odd += i

print("Even sum is {}".format(sum_even))
print("Odd sum is {}".format(sum_odd))
print("Sum of even numbers:", sum_even)
print("Sum of odd numbers:", sum_odd)
```

```python
# while loop

# i = 0
# while(i <= 10):
#     print(i)
#     i += 1

i = 0
even_sum = 0
odd_sum = 0
while(i <= 10):
    if(i % 2 == 0):
        even_sum += i
    else:
        odd_sum += i
    i += 1

print(even_sum, odd_sum)
```

```python
# beak statement and continue statement

x = 1
while(x < 7):
    if(x == 4):
        break
    print(x)
    x += 1

x = 0
while(x < 7):
    x += 1
    if(x == 4):
        continue
    print(x)
    
```

```python
# reverse a number
n = 1234
rev = 0

while(n > 0):
    d = n % 10
    rev = rev * 10 + d
    n = n //10

print(rev)
```
