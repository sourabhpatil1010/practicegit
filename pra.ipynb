```python
def factorial(n):
    if(n == 0 or n == 1):
        return 1
    else:
        return n * factorial(n - 1)
```

```python
def fact(n):
    if(n == 0 or n == 1):
        return 1
    
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

```python
print(factorial(5))
print(fact(5))
```
