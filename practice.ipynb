```python
# variables and data types
a = 10
print(type(a)) # int
print(a)

s = "string"
print(type(s)) # str
print(s)
```

```python
# integer
x = 10
print(type(x))

print(x+10)
print(x-5)
print(x*2)
print(x//2)
print(x%3) # modulus operator
print(x**2) # exponentiation operator

a = 10
b = 20
print(a+b)
```

```python
# Floating point numbers
y = 3.14
print(type(y))
print(y)

# Type conversion
z = int(y) # converts float to int
print(z)
w = float(z) # converts int to float
print(w)
```

```python
# Boolean
a1 = True
b1 = False
print(type(a1))
print(a1 and b1) # logical AND
print(a1 or b1) # logical OR
print(not a1) # logical NOT
```

```python
# STRING
s1 = "Hello"
s2 = 'World'
print(s1)
print(type(s1))
print(s2)

print(s1 + " " + s2) # string concatenation
print(s1 * 3) # string repetition
#s2 + 2 # TypeError: can only concatenate str (not "int") to str
print(s2 + str(2)) # converts int to str and concatenates
```

```python
# complex numbers
j = 1.0-2.3j
print(type(j))
print(j.real, j.imag)
```

```python
# dynamic typing and Strong typing
x = 10
print(x)
print(type(x)) # int
x = "string"
print(x)
print(type(x)) # str

# strong typing example
a = "hi"
print(a + "1")
# TypeError: can only concatenate str (not "int") to str
#print(a + 1) # TypeError: can only concatenate str (not "int") to str
print(a + str(1)) # converts int to str and concatenates
```

```python
# String formatting
name = "Alice"
a = 10
print("my num is - ", a)
print("my name is - ",name)

# f-string formatting
name = "Alice"
age = 30
print("my name is {} and my age is {}".format(name,age))

print("my name is {1} and my age is {0}".format(age, name)) # positional arguments
```

```python
c = 3.14
print(f"the value of c is {c}") # f-string formatting

a = input("Enter a number: ")
b = input("Enter another number: ")
print("The sum is: ", a + b) # concatenates as strings
print("The sum is: ", int(a) + int(b))
```
